```python
#listcollecttion of elements.
#it is denoted by [] with comma separated value
#orderd,indexed,mutable and hetrogenous in nature
l=[]
print(l)
print(type(l))
```

```python
l=[10,20,30,"hello",10.5,1000]
print(l)
print(type(l))
print(l[2]) #indexed
l[2]=3000 # muteable
print(l)
```

```python
# dictionry -- collection of elements
#denoted by {k:v} with kry: value pair
# keys must be unique value can be duplicate.
# ordered ,unidexed,muteble.
d={} # empty dictionory
print(d)
print(type(d))
```

```python
d={"name":"abc","course":"DA","fees":3000}
print(d)
print(type(d))
print(d["course"])
d["name"]="xyz"  # muteble and unindexed
print(d)
```

```python
#set -collection of unique element
#denoted by {} with  cvs.
# unoderd,unidexed,mutable and heterogenous in nature
s=set() # set constructor
print(s)
print(type(s))
```

```python
s={10,20,30,"hello",10.5,10,20,10,20,10}
print(s)
print(type(s))
s.add(1000) # mutable
print(s)
```

```python
# frozenset - collection od elements
# denoted by constructor -frozenset()
# unordered ,unindexed,immuteble and heterogenous in nature
f=frozenset()
print(f)
print(type(f))
```

```python
f=frozenset({10,20,30,"hello",10.5,10,20,10})
print(f)
print(type(f))
f.clear(100)
print(f)
```
