```python
# wap to print first and natural n number
# wap to print first n even number and first n odd number take n as a single input 
```

```python
n=int(input("enter any number: "))
sum_even=0
sum_odd=0
for i in range(1,2*n+1):
    if i%2==0:
        sum_even=sum_even+i
    else:
        sum_odd=sum_odd+i
print(sum_even)
print(sum_odd)
```
