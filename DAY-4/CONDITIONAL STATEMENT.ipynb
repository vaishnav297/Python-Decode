```python
# WAP TO CHECK  WHETRE A TRINGLE IS VAILID OR INVAILID IF IT IS VAILID CHECK IS IT EQUILATERAL,SCALEN,ISCOCELAN FOR VALIDITY OF 
# TRINGLE PLEASE CHECK SITE THEURAM
def check_triangle():
    try:
        # Taking input from the user
        a = float(input("Enter length of side 1: "))
        b = float(input("Enter length of side 2: "))
        c = float(input("Enter length of side 3: "))

        # 1. Check for valid triangle using Triangle Inequality Theorem
        if (a + b > c) and (a + c > b) and (b + c > a):
            print("\nThe triangle is VALID.")

            # 2. Check for triangle type
            if a == b == c:
                print("Type: Equilateral Triangle (All sides are equal).")
            elif a == b or b == c or a == c:
                print("Type: Isosceles Triangle (Two sides are equal).")
            else:
                print("Type: Scalene Triangle (No sides are equal).")
                
        else:
            print("\nThe triangle is INVALID. The sum of any two sides must be greater than the third side.")
            
    except ValueError:
        print("Please enter valid numerical values.")

# Run the function
check_triangle()
```

```python
assignment = float(input("Enter assignment score: "))
midterm = float(input("Enter midterm exam score: "))
final_exam = float(input("Enter final exam score: "))
attendance = float(input("Enter attendance percentage: "))

total = (assignment * 0.30) + (midterm * 0.30) + (final_exam * 0.30) + (attendance * 0.10)

if attendance >= 95:
    total += 5

if total >= 90:
    grade = "A"
elif total >= 80:
    grade = "B"
elif total >= 70:
    grade = "C"
elif total >= 60:
    grade = "D"
else:
    grade = "F"

print("Final Grade:", grade)
```

```python
# Take the year input from the user
year = int(input("Enter a year: "))

# Check for leap year conditions:
# 1. Divisible by 4 and not divisible by 100 OR
# 2. Divisible by 400
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("Leap year")
else:
    print("Not a leap year")
```
